## 2. Load Prototype Fashion Data & Evaluate CNN

```python
# Load persisted Day 6 model
model = tf.keras.models.load_model('../models/catalog_cnn_v6.keras')

fashion_mnist = tf.keras.datasets.fashion_mnist
_, (test_images, test_labels) = fashion_mnist.load_data()
test_images = np.expand_dims(test_images, -1).astype('float32') / 255.0

class_names = ['T-shirt/top', 'Trouser', 'Pullover', 'Dress', 'Coat', 'Sandal', 'Shirt', 'Sneaker', 'Bag', 'Ankle boot']

y_pred = np.argmax(model.predict(test_images), axis=1)
cm = confusion_matrix(test_labels, y_pred)

plt.figure(figsize=(10, 8))
sns.heatmap(cm, annot=True, fmt='d', cmap='Blues', xticklabels=class_names, yticklabels=class_names)
plt.ylabel('Actual Product')
plt.xlabel('Predicted Auto-Tag')
plt.title('CNN Confusion Matrix: Where is the model confused?')
plt.show()
```
